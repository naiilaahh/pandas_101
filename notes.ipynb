### Pandas Datastructures
- Series - single dimensional datastructure. Similar to a list of single column (homogeneous of the same data type)
- Dataframe - multidimensional table (heterogeneous of many datatypes)

```python
ser1 = pd.Series([10, 20, 30, 40], index=["A", "B", "C", "D"])
print(ser1)
```

```python
# creating aseries from a preexisting dictionary
d = {
    "b": 1,
    "a": 0,
    "c": 2,
    "d": 10
}

ser2 = pd.Series(d)
ser2.iloc[3]
```

```python
d1 = {
    "a":0.0,
    "b":1.0,
    "c":2.0
}

ser3 = pd.Series(d1)
print(ser3)
```

```python
calories = {
    'day1':700,
    'day2':2500,
    'day3':1200
}
# use indexes to choose certain calues for a series
ser4 = pd.Series(calories, index=('day1', 'day2'))
print(ser4)
```

```python
ser5 = pd.Series(5.0, index=["a", "b", "c"])

print(ser5)
```

# Sort using pandas

- `.sort_values()`

```python
data = {'Name': ['Jane', 'Princi', 'James', 'Fadi'],
        'Height': [5.1, 6.2, 5.1, 5.2],
        'Qualification': ['Msc', 'MA', 'Msc', 'Msc'],
       'Score 1' : [56,86,77,45],
       'Score 2' : [50,96,60,30]}

grades_df = pd.DataFrame(data)
```

```python
grades_df.head()
```

```python
grades_df = grades_df.assign(address=['NYC', 'NJ', 'CA', 'PA'])
```

```python
grades_df.head()
```

```python
print(grades_df.sort_values(by='Score 1'))
```

```python
print(grades_df.sort_values(by='Score 1', ascending=False))
```

### Slicing a dataframe with .iloc[] & .loc[]
-   `df.loc[ row:row , colName:colName ]` - label based indexing
-   `df.iloc[row:row, [colNumber:colNumber]` - index based location

#### .iloc

```python
# Initializing the nested list with Data set
employee_list = [['James', 36, 75, 5428000],
               ['Villers', 38, 74, 3428000],
               ['VKole', 31, 70, 8428000],
               ['Smith', 34, 80, 4428000],
               ['Gayle', 40, 100, 4528000],
               ['Rooter', 33, 72, 7028000],
               ['Peterson', 42, 85, 2528000],
               ['John', 41, 85, 1528000],

]
```

```python
e_df = pd.DataFrame(employee_list, columns=['Name', 'Age', 'Weight', 'Salary'])

e_df.head()
```

```python
# Grabs the first row
e_df.iloc[:1]
```

```python
e_df.iloc[0:4]
```

```python
# Grabs all rows, but only first 3 columns
e_df.iloc[:, 0:3]
```

```python
# Grab first 2 rows and first 2 columns

e_df.iloc[:2, :2]
```

#### .loc

```python
basketball = {'team': ['A', 'B', 'C', 'D', 'E', 'F', 'G', 'H'],
                   'points': [18, 22, 19, 14, 14, 11, 20, 28],
                   'assists': [5, 7, 7, 9, 12, 9, 9, 4],
                   'rebounds': [11, 8, 10, 6, 6, 5, 9, 12],
                   'steals': [4, 3, 3, 2, 5, 4, 3, 8],
                   'blocks': [1, 0, 0, 3, 2, 2, 1, 5]}
```

```python
ball = pd.DataFrame(basketball)
```

```python
ball.head()
```

```python
# Locate all rows, and colums team and rebounds

ball.loc[:, ['team', 'rebounds']]
```

```python
# select values from rows 3 to 6 and show the team name

ball.loc[3:6, ['team']]
```

# Pandas Basics

- Pandas is a Python library for data analysis and manipulation.
- Built on NumPy (math operations) and Matplotlib (visualization).

```python
import numpy as np
import pandas as pd
```

```python
print(f'Pandas: {pd.__version__}')
print(f'Numpy: {np.__version__}')
```

### Pandas Datastructures
- Series - single dimensional datastructure. Similar to a list of single column (homogeneous of the same data type)
- Dataframe - multidimensional table (heterogeneous of many datatypes)

```python
ser1 = pd.Series([10, 20, 30, 40], index=["A", "B", "C", "D"])
print(ser1)
```

```python
# creating aseries from a preexisting dictionary
d = {
    "b": 1,
    "a": 0,
    "c": 2,
    "d": 10
}

ser2 = pd.Series(d)
ser2.iloc[3]
```

```python
d1 = {
    "a":0.0,
    "b":1.0,
    "c":2.0
}

ser3 = pd.Series(d1)
print(ser3)
```

```python
calories = {
    'day1':700,
    'day2':2500,
    'day3':1200
}
# use indexes to choose certain calues for a series
ser4 = pd.Series(calories, index=('day1', 'day2'))
print(ser4)
```

```python
ser5 = pd.Series(5.0, index=["a", "b", "c"])

print(ser5)
```

### Data Frames
- indexes (each record)
- columns (each variable)


```python
fruit_data = {
    'apple_sales': [3, 2, 0, 2, 9, 4, 7, 2],
    'orange_sales': [0, 3, 2, 7, 7, 8, 3, 0 ]
}

purchases = pd.DataFrame(fruit_data, index=['Day1', 'Day2', 'Day3', 'Day4', 'Day5', 'Day6', 'Day7', 'Day8'])

print(purchases)
```

```python
rows, cols = purchases.shape
print(f'Rows: {rows}')
print(f'Cols: {cols}')
```

```python
# to view the top number of elements, by default gives 5 elements
print(purchases.head())
```

```python
# to view the bottom number of elements, by default 5

print(purchases.tail())
```

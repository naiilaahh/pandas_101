```python
import numpy as np
import pandas as pd
```

```python
data = {'Name': ['Jane', 'Princi', 'James', 'Fadi'],
        'Height': [5.1, 6.2, 5.1, 5.2],
        'Qualification': ['Msc', 'MA', 'Msc', 'Msc'],
        'Score 1' : [56,86,77,45],
        'Score 2' : [50,96,60,30]}
```

```python
data_df = pd.DataFrame(data)
data_df.head()
```

```python
address = ['NYC', 'NJ', 'CA', 'PA']
data_df['Address'] = address
```

```python
data_df.head()
```

```python
# creating a new column using an old column(S)
print('Before Column Add:')
data_df.head()
```

```python
# Add column using old columns
data_df['Total Score'] = data_df['Score 1'] + data_df['Score 2']

print('After column Added:')
data_df.head()
```

```python
data_df['Avg Score'] = data_df['Total Score'] / 2
data_df.head()
```

```python
df2 = pd.DataFrame(data, columns=['Score 1', 'Score 2', 'Name', 'Qualification'])
print(df2)
```
